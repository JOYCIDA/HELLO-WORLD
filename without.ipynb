```python
import math

dataset = [
    {"Study Hours": "High",   "Attendance": "Good", "Pass": "Yes"},
    {"Study Hours": "High",   "Attendance": "Poor", "Pass": "Yes"},
    {"Study Hours": "Medium", "Attendance": "Good", "Pass": "Yes"},
    {"Study Hours": "Low",    "Attendance": "Good", "Pass": "No"},
    {"Study Hours": "Low",    "Attendance": "Poor", "Pass": "No"},
    {"Study Hours": "Medium", "Attendance": "Poor", "Pass": "No"},
]

def entropy(data):
    n = len(data)
    counts = {}
    for row in data:
        label = row["Pass"]
        counts[label] = counts.get(label, 0) + 1
    e = 0
    for count in counts.values():
        p = count / n
        e -= p * math.log2(p)
    return e

def info_gain(data, feature):
    base = entropy(data)
    values = set(row[feature] for row in data)
    weighted = 0
    for v in values:
        subset = [row for row in data if row[feature] == v]
        weighted += (len(subset) / len(data)) * entropy(subset)
    return base - weighted

for feat in ["Study Hours", "Attendance"]:
    print(f"Information Gain ({feat}): {info_gain(dataset, feat):.4f}")

best = max(["Study Hours", "Attendance"], key=lambda f: info_gain(dataset, f))
print(f"\nBest split: {best}")
```

```python
points = [("A",1,"Fail"),("B",2,"Fail"),("C",3,"Fail"),
          ("D",6,"Pass"),("E",7,"Pass"),("F",8,"Pass")]
new_point = 5

distances = []
for student, hours, result in points:
    d = abs(new_point - hours)
    distances.append((d, student, result))
distances.sort()

for k in [1, 3, 5]:
    neighbors = distances[:k]
    votes = {"Pass": 0, "Fail": 0}
    for d, s, r in neighbors:
        votes[r] += 1
    print(f"K={k}: {max(votes, key=votes.get)}")
```

```python
import math

points = {"A":(1,1), "B":(2,1), "C":(4,3), "D":(5,4)}

def distance(p1, p2):
    return math.sqrt((p1[0]-p2[0])**2 + (p1[1]-p2[1])**2)

def centroid(pts):
    return (sum(p[0] for p in pts)/len(pts), sum(p[1] for p in pts)/len(pts))

c1, c2 = (1,1), (5,4)

for iteration in range(1, 3):
    cluster1, cluster2 = [], []
    for name, pt in points.items():
        if distance(pt, c1) <= distance(pt, c2):
            cluster1.append(pt)
        else:
            cluster2.append(pt)
    new_c1, new_c2 = centroid(cluster1), centroid(cluster2)
    print(f"Iter {iteration}: C1={new_c1}, C2={new_c2}")
    if new_c1 == c1 and new_c2 == c2:
        print("Converged!")
        break
    c1, c2 = new_c1, new_c2
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]
n = len(x)
mean_x = sum(x) / n
mean_y = sum(y) / n
num = sum((x[i]-mean_x)*(y[i]-mean_y) for i in range(n))
den = sum((x[i]-mean_x)**2 for i in range(n))
m = num / den
b = mean_y - m * mean_x
print(f"Equation: y = {m}x + {b}")
for h in [6, 7]:
    print(f"{h} hours → {m*h+b} marks")
```
