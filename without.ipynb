```python
# ============================================================
# Exercise 1 — Linear Regression: Delivery Time Prediction
# ============================================================


# STARTER CODE - Run first
distance = [2, 4, 5, 7, 8, 10, 12, 13]
time = [18, 28, 35, 44, 52, 61, 70, 76]
n = len(distance)
print('Number of deliveries:', n)


# Task 1
# I need 4 values to calculate the regression line later.
# sum_x  = total of all distances
# sum_y  = total of all delivery times
# sum_xy = each distance multiplied by its time, all added up
# sum_xx = each distance squared, all added up
# I used a single loop to compute all four at once to keep it clean.

sum_x, sum_y, sum_xy, sum_xx = 0, 0, 0, 0
for i in range(n):
    sum_x  += distance[i]
    sum_y  += time[i]
    sum_xy += distance[i] * time[i]
    sum_xx += distance[i] ** 2

print("sum_x:", sum_x, "| sum_y:", sum_y, "| sum_xy:", sum_xy, "| sum_xx:", sum_xx)


# Task 2
# Using the 4 sums from Task 1, I applied the linear regression formulas
# to find m (slope) and b (intercept).
# m tells me how much the delivery time increases per extra kilometre.
# b is the base time even when distance is 0 (fixed overhead like loading time).

m = (n * sum_xy - sum_x * sum_y) / (n * sum_xx - sum_x ** 2)
b = (sum_y - m * sum_x) / n

print("Slope m:", round(m, 2))
print("Intercept b:", round(b, 2))

# m = 5.26 means: for every extra km, the delivery takes about 5.26 more minutes.
# So if I compare a 3 km delivery to a 4 km delivery, I expect ~5 extra minutes.


# Task 3
# I used the equation predicted_time = m * distance + b to estimate
# delivery times for 3 distances not in the original dataset.
# This is where linear regression becomes useful — predicting unseen values.

for d in [6, 9, 15]:
    predicted_time = m * d + b
    print(f"Distance {d} km → Predicted time: {round(predicted_time, 1)} mins")


# Task 4
# For each of the 8 deliveries, I calculated the error:
# error = actual time - predicted time
# A negative error means the model overestimated (predicted too high).
# A positive error means it underestimated (predicted too low).
# Then I calculated the MAE, which is the average size of the errors
# regardless of direction (that's why I used abs()).

errors = []
for i in range(n):
    predicted = m * distance[i] + b
    error = time[i] - predicted
    errors.append(error)
    print(f"Delivery {i+1}: actual={time[i]}, predicted={round(predicted,1)}, error={round(error,2)}")

mae = sum(abs(e) for e in errors) / n
print(f"MAE: {round(mae, 2)}")


# Task 5
# My MAE is 0.84 minutes.
# That means on average, my model is off by less than 1 minute per delivery.
# For a delivery company, I think this is very acceptable — being wrong by
# under a minute on a delivery prediction is not going to cause any real problem.
# If the MAE were something like 10 or 15 minutes, that would be a problem
# because customers would be waiting much longer than expected.
# 0.84 is a strong result given we only had 8 data points to train on.

print(f"MAE = {round(mae, 2)} mins — model is accurate enough for real use.")


# Task 6
# I added a 9th delivery: distance=10 km, time=80 mins.
# That delivery took 80 minutes but the model predicted around 60 mins for 10 km,
# so it is clearly an outlier — something unusual must have happened (traffic, delay).
# After recalculating with 9 points, both m and b changed.

distance2 = distance + [10]
time2     = time + [80]
n2        = len(distance2)

sx  = sum(distance2)
sy  = sum(time2)
sxy = sum(distance2[i] * time2[i] for i in range(n2))
sxx = sum(distance2[i] ** 2       for i in range(n2))

m2 = (n2 * sxy - sx * sy) / (n2 * sxx - sx ** 2)
b2 = (sy - m2 * sx) / n2

print(f"New m: {round(m2, 2)} (was {round(m, 2)})")
print(f"New b: {round(b2, 2)} (was {round(b, 2)})")

# m went from 5.26 to 5.63 — the slope got steeper because of the outlier.
# b dropped from 7.89 to 7.12.
# This shows that even a single unusual data point can pull the regression line
# in a different direction. If the delivery company keeps recording late deliveries
# without flagging them as outliers, the model will gradually become less accurate
# for normal deliveries.
```

```python

# ============================================================
# Exercise 2 — K-Nearest Neighbors (KNN): Fruit Classifier
# ============================================================


# STARTER CODE - Run first
fruits = [
    {'id':1, 'weight':150, 'diameter':7.0, 'label':'Apple'},
    {'id':2, 'weight':165, 'diameter':7.4, 'label':'Apple'},
    {'id':3, 'weight':170, 'diameter':7.2, 'label':'Apple'},
    {'id':4, 'weight':140, 'diameter':6.8, 'label':'Apple'},
    {'id':5, 'weight':200, 'diameter':8.5, 'label':'Orange'},
    {'id':6, 'weight':210, 'diameter':8.8, 'label':'Orange'},
    {'id':7, 'weight':195, 'diameter':8.3, 'label':'Orange'},
    {'id':8, 'weight':185, 'diameter':8.0, 'label':'Orange'},
]
unknown = {'weight': 175, 'diameter': 7.8}
K = 3
print('Dataset loaded. Unknown fruit:', unknown)


# Task 1
# The first thing I need is a way to measure how far apart two fruits are.
# I used the Euclidean distance formula which works in 2D space —
# here the two dimensions are weight and diameter.
# I tested it on Fruit 1 to make sure the function works correctly.
# The expected result is around 25.0 so if I get that, my function is good.

def euclidean(f, u):
    return ((f['weight'] - u['weight'])**2 + (f['diameter'] - u['diameter'])**2) ** 0.5

test_distance = round(euclidean(fruits[0], unknown), 2)
print('Distance to Fruit 1:', test_distance)


# Task 2
# Now I calculate the distance from the unknown fruit to every fruit in the dataset.
# I store each result as a tuple (distance, label, id) so I can sort them easily.
# Sorting by distance gives me the closest fruits first,
# which is exactly what KNN needs to find the nearest neighbours.

all_distances = []
for f in fruits:
    d = euclidean(f, unknown)
    all_distances.append((round(d, 2), f['label'], f['id']))

all_distances.sort()

print('All distances sorted (smallest to largest):')
for d, label, fruit_id in all_distances:
    print(f'  Fruit {fruit_id}: distance={d}, label={label}')


# Task 3
# With K=3, I take the 3 closest fruits and let them vote.
# Whichever label appears most often wins — that is the KNN majority vote.
# I expected Fruit 3 to be the closest since it is the most similar
# to the unknown fruit (weight=170, diameter=7.2 vs unknown weight=175, diameter=7.8).

K = 3
neighbors = all_distances[:K]

votes = {'Apple': 0, 'Orange': 0}
for d, label, fruit_id in neighbors:
    votes[label] += 1

print(f'K = {K} nearest neighbours:')
for d, label, fruit_id in neighbors:
    print(f'  Fruit {fruit_id} → {label} (distance: {d})')

print('Votes:', votes)
print('Prediction for unknown fruit:', max(votes, key=votes.get))

# Result: 2 Apples vs 1 Orange → the unknown fruit is classified as Apple.
# This makes sense because weight=175g and diameter=7.8cm are closer
# to the Apple range (140-170g, 6.8-7.4cm) than to Oranges (185-210g, 8.0-8.8cm).


# Task 4
# I tried K=5 to see if using more neighbours changes the prediction.
# With 5 neighbours I now include more fruits further away,
# but the majority is still Apple so the answer does not change.
# The answer stayed the same because the unknown fruit sits physically
# closer to the Apple cluster — adding more neighbours does not change that fact.

K = 5
neighbors5 = all_distances[:K]

votes5 = {'Apple': 0, 'Orange': 0}
for d, label, fruit_id in neighbors5:
    votes5[label] += 1

print(f'K = {K} nearest neighbours:')
for d, label, fruit_id in neighbors5:
    print(f'  Fruit {fruit_id} → {label} (distance: {d})')

print('Votes:', votes5)
print('Prediction with K=5:', max(votes5, key=votes5.get))


# Task 5
# K=7 means I am using almost the entire dataset (7 out of 8 fruits).
# At this point the vote is 4 Apples vs 3 Oranges — still Apple, but very close.
# The problem with K=7 here is that the dataset only has 4 Apples and 4 Oranges.
# When K gets close to the total number of points, the model stops looking
# at what is actually nearby and starts reflecting the overall class balance instead.
# If the dataset had 5 Oranges and 3 Apples, the result would flip to Orange
# just because of that imbalance — not because the fruit is genuinely more Orange.
# That is why choosing K too large is risky with a small balanced dataset.

K = 7
neighbors7 = all_distances[:K]

votes7 = {'Apple': 0, 'Orange': 0}
for d, label, fruit_id in neighbors7:
    votes7[label] += 1

print(f'K = {K} nearest neighbours:')
for d, label, fruit_id in neighbors7:
    print(f'  Fruit {fruit_id} → {label} (distance: {d})')

print('Votes:', votes7)
print('Prediction with K=7:', max(votes7, key=votes7.get))


# Task 6
# I noticed a problem with my original distance function:
# weight goes from 140 to 210 (a range of 70),
# while diameter only goes from 6.8 to 8.8 (a range of 2).
# This means a 10g difference in weight adds much more to the distance
# than a 1cm difference in diameter, even if diameter is just as important.
# To fix this I normalised both features to a 0-1 scale using min-max normalisation.
# Now both features contribute equally to the distance calculation.

# Step 1: Find min and max for each feature
min_weight   = min(f['weight']   for f in fruits)
max_weight   = max(f['weight']   for f in fruits)
min_diameter = min(f['diameter'] for f in fruits)
max_diameter = max(f['diameter'] for f in fruits)

print(f'Weight range:   {min_weight} – {max_weight}')
print(f'Diameter range: {min_diameter} – {max_diameter}')

# Step 2: Normalised euclidean distance function
# I apply the same formula but on the scaled values instead of raw ones
def euclidean_normalised(f, u):
    norm_w_f = (f['weight']   - min_weight)   / (max_weight   - min_weight)
    norm_d_f = (f['diameter'] - min_diameter) / (max_diameter - min_diameter)
    norm_w_u = (u['weight']   - min_weight)   / (max_weight   - min_weight)
    norm_d_u = (u['diameter'] - min_diameter) / (max_diameter - min_diameter)
    return ((norm_w_f - norm_w_u)**2 + (norm_d_f - norm_d_u)**2) ** 0.5

# Step 3: Recalculate all distances using normalised values
norm_distances = []
for f in fruits:
    d = euclidean_normalised(f, unknown)
    norm_distances.append((round(d, 4), f['label'], f['id']))

norm_distances.sort()

print('\nNormalised distances sorted:')
for d, label, fruit_id in norm_distances:
    print(f'  Fruit {fruit_id}: distance={d}, label={label}')

# Step 4: K=3 prediction with normalised distances
K = 3
votes_norm = {'Apple': 0, 'Orange': 0}
for d, label, fruit_id in norm_distances[:K]:
    votes_norm[label] += 1

print(f'\nK=3 neighbours after normalisation:')
for d, label, fruit_id in norm_distances[:K]:
    print(f'  Fruit {fruit_id} → {label} (distance: {d})')

print('Votes:', votes_norm)
print('Prediction after normalisation:', max(votes_norm, key=votes_norm.get))

# After normalisation the ranking changed slightly — Fruit 8 (Orange) moved up
# to become the closest neighbour instead of being second.
# But the final prediction is still Apple (2 Apple vs 1 Orange).
# What this tells me is that when features have very different scales,
# the larger-scale feature dominates unfairly.
# Normalising is important to make sure KNN judges distance fairly
# across all features, not just the ones with bigger numbers.
```

```python

# ============================================================
# Exercise 3 — Decision Tree: Loan Approval
# ============================================================


# STARTER CODE - Run first
# The gini() function is already given — I just need to call it correctly.
# Gini impurity measures how mixed a group is.
# A value of 0.0 means all decisions in that group are the same (pure).
# A value of 0.5 means the group is perfectly split between Approve and Reject.

data = [
    {'income':'High',   'credit':'Good', 'employed':'Yes', 'decision':'Approve'},
    {'income':'High',   'credit':'Good', 'employed':'No',  'decision':'Approve'},
    {'income':'Medium', 'credit':'Good', 'employed':'Yes', 'decision':'Approve'},
    {'income':'Low',    'credit':'Good', 'employed':'Yes', 'decision':'Approve'},
    {'income':'Medium', 'credit':'Poor', 'employed':'Yes', 'decision':'Reject'},
    {'income':'High',   'credit':'Poor', 'employed':'No',  'decision':'Reject'},
    {'income':'Low',    'credit':'Poor', 'employed':'Yes', 'decision':'Reject'},
    {'income':'Low',    'credit':'Poor', 'employed':'No',  'decision':'Reject'},
]

def gini(group):
    n = len(group)
    if n == 0: return 0
    approve = sum(1 for d in group if d['decision'] == 'Approve') / n
    reject = 1 - approve
    return 1 - (approve**2 + reject**2)

print('Root Gini (whole dataset):', round(gini(data), 3))
# Root Gini = 0.5 makes complete sense here.
# The dataset has exactly 4 Approve and 4 Reject — perfectly balanced.
# A perfectly balanced dataset gives the maximum impurity score of 0.5,
# which means the tree has no useful information yet at the root level.


# Task 1
# I split the dataset by Credit History into two groups: Good and Poor.
# My goal is to see whether knowing someone's credit history
# helps the tree separate Approve from Reject decisions.
# I calculate Gini for each group, then combine them into a weighted average
# based on how many applicants are in each group.

good = [d for d in data if d['credit'] == 'Good']
poor = [d for d in data if d['credit'] == 'Poor']

gini_good = gini(good)
gini_poor = gini(poor)
weighted_credit = (len(good) / 8) * gini_good + (len(poor) / 8) * gini_poor

print(f'Group Good  ({len(good)} applicants): Gini = {round(gini_good, 3)}')
print(f'Group Poor  ({len(poor)} applicants): Gini = {round(gini_poor, 3)}')
print(f'Weighted Gini for Credit History split: {round(weighted_credit, 3)}')

# Both groups came out with Gini = 0.0 — that is a perfect result.
# Every single applicant with Good credit was Approved,
# and every single applicant with Poor credit was Rejected.
# A weighted Gini of 0.0 means this split completely eliminates impurity.
# I already have a feeling this will be the winning feature.


# Task 2
# I now test Income as the splitting feature.
# Income has three possible values (High, Medium, Low),
# so I need three separate groups and a weighted average across all three.

high   = [d for d in data if d['income'] == 'High']
medium = [d for d in data if d['income'] == 'Medium']
low    = [d for d in data if d['income'] == 'Low']

gini_high   = gini(high)
gini_medium = gini(medium)
gini_low    = gini(low)

weighted_income = (
    (len(high)   / 8) * gini_high   +
    (len(medium) / 8) * gini_medium +
    (len(low)    / 8) * gini_low
)

print(f'Group High   ({len(high)} applicants): Gini = {round(gini_high, 3)}')
print(f'Group Medium ({len(medium)} applicants): Gini = {round(gini_medium, 3)}')
print(f'Group Low    ({len(low)} applicants): Gini = {round(gini_low, 3)}')
print(f'Weighted Gini for Income split: {round(weighted_income, 3)}')

# Weighted Gini for Income = 0.458 — much worse than Credit History.
# Each income group still contains a mix of Approve and Reject,
# which means Income alone cannot cleanly separate the two outcomes.
# A High income applicant can still be Rejected, and a Low income applicant
# can still be Approved, so this feature does not give the tree clear guidance.


# Task 3
# Last feature to test: Employment status (Yes or No).
# I split into two groups the same way I did for Credit History in Task 1.

employed_yes = [d for d in data if d['employed'] == 'Yes']
employed_no  = [d for d in data if d['employed'] == 'No']

gini_yes = gini(employed_yes)
gini_no  = gini(employed_no)
weighted_employment = (len(employed_yes) / 8) * gini_yes + (len(employed_no) / 8) * gini_no

print(f'Group Employed=Yes ({len(employed_yes)} applicants): Gini = {round(gini_yes, 3)}')
print(f'Group Employed=No  ({len(employed_no)} applicants): Gini = {round(gini_no, 3)}')
print(f'Weighted Gini for Employment split: {round(weighted_employment, 3)}')

# Employment gives 0.467 — slightly worse than Income and far worse than Credit.
# Being employed or not does not cleanly predict whether a loan gets approved.
# Looking at the data, there are both Approve and Reject cases for employed people,
# so Employment is not a reliable separator on its own.


# Task 4
# I compare all three weighted Gini scores side by side.
# The lowest score wins — that feature becomes the root of the decision tree.
# I used a dictionary to store the scores and min() to find the best one cleanly.

scores = {
    'Credit History': weighted_credit,
    'Income':         weighted_income,
    'Employment':     weighted_employment,
}

print('Weighted Gini scores:')
for feature, score in scores.items():
    print(f'  {feature}: {round(score, 3)}')

best_feature = min(scores, key=scores.get)
print(f'\nBest root split: {best_feature} (lowest Gini = {round(scores[best_feature], 3)})')

# Credit History wins by a large margin with a Gini of 0.0.
# A lower Gini is better because it means the split produces purer groups —
# groups where all (or most) applicants share the same decision.
# The tree wants to ask the question that gives the most useful answer,
# and Credit History does exactly that: it perfectly divides Approve from Reject
# without any confusion or overlap.


# Task 5
# Since Credit History already gives Gini = 0.0, both branches are pure.
# But I still check the Poor branch to confirm it needs no further splitting,
# and also to understand which feature would come second if it were needed.

poor_branch = [d for d in data if d['credit'] == 'Poor']

print(f'Poor branch size: {len(poor_branch)}')
print('Applicants in Poor branch:')
for d in poor_branch:
    print(f'  Income={d["income"]}, Employed={d["employed"]}, Decision={d["decision"]}')

# Test Income on poor branch
income_values = set(d['income'] for d in poor_branch)
weighted_income_poor = 0
for val in income_values:
    subset = [d for d in poor_branch if d['income'] == val]
    weighted_income_poor += (len(subset) / len(poor_branch)) * gini(subset)

# Test Employment on poor branch
employed_values = set(d['employed'] for d in poor_branch)
weighted_employed_poor = 0
for val in employed_values:
    subset = [d for d in poor_branch if d['employed'] == val]
    weighted_employed_poor += (len(subset) / len(poor_branch)) * gini(subset)

print(f'\nWeighted Gini (Income)     on Poor branch: {round(weighted_income_poor, 3)}')
print(f'Weighted Gini (Employment) on Poor branch: {round(weighted_employed_poor, 3)}')
print()
print('Both give Gini = 0.0 because the Poor branch is already pure (all Reject).')
print('No second-level split is needed — Credit History alone is sufficient.')

# As expected, both Income and Employment give 0.0 on the Poor branch
# because every applicant there is already Reject — there is nothing left to separate.
# This confirms that the decision tree only needs one level for this dataset.


# Task 6
# I wrote the predict() function using if/else to follow the tree structure.
# The logic is very simple here: the tree only asks one question.
# If credit is Good → Approve. If credit is Poor → Reject.
# Income and Employment are not used because they were not needed.

def predict(income, credit, employed):
    # Level 1: split on Credit History (best root feature, Gini = 0.0)
    if credit == 'Good':
        return 'Approve'
    else:
        return 'Reject'

test_cases = [
    ('High',   'Poor', 'Yes'),
    ('Low',    'Good', 'No'),
    ('Medium', 'Poor', 'No'),
]

print('Predictions for new applicants:')
for income, credit, employed in test_cases:
    result = predict(income, credit, employed)
    print(f'  Income={income}, Credit={credit}, Employed={employed} → {result}')

# The results look correct to me:
# (High, Poor, Yes) → Reject — Poor credit overrides everything else.
# (Low, Good, No)   → Approve — Good credit is enough even without a job.
# (Medium, Poor, No)→ Reject — Poor credit again, no surprise.


# Task 7
# I tested the function on a profile the tree has never seen before:
# Medium income, Good credit, Not employed.
# The tree gives Approve because credit = Good — that is its only rule.
# But this profile does not appear anywhere in the training data,
# so I wanted to check whether the dataset even agrees with that prediction.

income_test   = 'Medium'
credit_test   = 'Good'
employed_test = 'No'

tree_prediction = predict(income_test, credit_test, employed_test)
print(f'predict({income_test}, {credit_test}, {employed_test}) → {tree_prediction}')

print('\nSearching dataset for exact match (Medium, Good, No)...')
match_found = False
for d in data:
    if d['income'] == income_test and d['credit'] == credit_test and d['employed'] == employed_test:
        print(f'  Match found: Decision = {d["decision"]}')
        match_found = True

if not match_found:
    print('  No exact match found in the dataset.')

# No match found — this exact combination was never in the training data.
# The tree still answers Approve, but it is basically applying the only rule it knows.
# This reveals a real limit of simple decision trees —
# they can only work with what they have seen during training.
# If an edge case was never in the data, the tree applies the closest rule it knows,
# which may or may not be the right answer in reality.
# A bank using this tree would need more data to handle unusual profiles confidently.
```

```python


# ============================================================
# Exercise 4 — K-Means Clustering: Wi-Fi Hotspot Zones
# ============================================================


# STARTER CODE - Run first
# The dataset has 6 hotspots spread across a map using (x, y) coordinates.
# The goal is to group them into 3 maintenance zones using K-Means.
# I also need to choose my own starting centroids — they are not given this time.

points = {
    'P1':(1,1), 'P2':(1,3), 'P3':(2,2),
    'P4':(8,1), 'P5':(9,2), 'P6':(8,3)
}

# My chosen starting centroids: P1(1,1), P5(9,2), P6(8,3)
# I picked these three because they are spread across different parts of the map.
# P1 is on the far left, P5 and P6 are on the right side.
# Starting with centroids that are already far apart gives each zone
# a fair territory from the beginning and avoids the algorithm
# getting stuck in a poor grouping from the start.
centroids = {
    'Zone1': (1, 1),
    'Zone2': (9, 2),
    'Zone3': (8, 3),
}

def distance(a, b):
    return ((a[0]-b[0])**2 + (a[1]-b[1])**2) ** 0.5

print('Hotspots:', points)
print('Starting centroids:', centroids)


# Task 1
# I replaced the ? values with my chosen starting points and justified the choice above.
# The key idea is to spread the centroids out so each one covers a different region.
# If I had chosen P1, P2, P3 — all from the left side — the right hotspots would
# all pile into one zone and the result would be meaningless geographically.

print('Centroids chosen and justified — see comments above.')
print('Zone1:', centroids['Zone1'], '← left area (P1)')
print('Zone2:', centroids['Zone2'], '← right area (P5)')
print('Zone3:', centroids['Zone3'], '← right area (P6)')


# Task 2
# In Iteration 1, I loop through every hotspot and calculate its distance
# to all 3 centroids. The hotspot gets assigned to whichever zone is closest.
# I store results in a dictionary so I can reuse the assignments in later tasks.

print('--- Iteration 1: Assigning hotspots to zones ---')

assignments_iter1 = {}

for name, pt in points.items():
    best_zone = None
    best_dist = float('inf')
    for zone, c in centroids.items():
        d = distance(pt, c)
        if d < best_dist:
            best_dist = d
            best_zone = zone
    assignments_iter1[name] = (best_zone, round(best_dist, 2))
    print(f'{name} → {best_zone} (distance: {round(best_dist, 2)})')

# P1, P2, P3 all went to Zone1 as expected — they are on the left side of the map.
# P4 and P5 went to Zone2, and P6 went to Zone3.
# This initial grouping already looks geographically reasonable.


# Task 3
# After the first assignment, I recalculate the centroid of each zone.
# The new centroid is simply the average x and average y of all points in that zone.
# This step moves the centroid to the true centre of its current members.

print('--- Iteration 1: Updating centroids ---')

zones_iter1 = {}
for name, (zone, d) in assignments_iter1.items():
    zones_iter1.setdefault(zone, []).append(points[name])

centroids_iter1 = {}
for zone, pts in zones_iter1.items():
    new_x = round(sum(p[0] for p in pts) / len(pts), 2)
    new_y = round(sum(p[1] for p in pts) / len(pts), 2)
    centroids_iter1[zone] = (new_x, new_y)

for zone, c in centroids_iter1.items():
    members = [n for n, (z, d) in assignments_iter1.items() if z == zone]
    print(f'{zone}: members={members} → new centroid = {c}')

# The centroids have moved from their starting positions.
# Zone1 shifted from (1,1) to (1.33, 2.0) because it now represents
# the average of P1, P2, and P3 rather than just the starting point P1.
# Zone2 and Zone3 also adjusted slightly to better reflect their members.


# Task 4
# I run a second iteration using the updated centroids from Task 3.
# I also check whether any hotspot switched zones compared to Iteration 1.
# If a hotspot changes zone, it means the centroid movement affected its assignment.

print('--- Iteration 2: Reassigning hotspots ---')

assignments_iter2 = {}

for name, pt in points.items():
    best_zone = None
    best_dist = float('inf')
    for zone, c in centroids_iter1.items():
        d = distance(pt, c)
        if d < best_dist:
            best_dist = d
            best_zone = zone
    assignments_iter2[name] = (best_zone, round(best_dist, 2))
    old_zone = assignments_iter1[name][0]
    changed = ' ← CHANGED ZONE!' if best_zone != old_zone else ''
    print(f'{name} → {best_zone} (distance: {round(best_dist, 2)}){changed}')

print()
print('--- Iteration 2: Updating centroids ---')

zones_iter2 = {}
for name, (zone, d) in assignments_iter2.items():
    zones_iter2.setdefault(zone, []).append(points[name])

centroids_iter2 = {}
for zone, pts in zones_iter2.items():
    new_x = round(sum(p[0] for p in pts) / len(pts), 2)
    new_y = round(sum(p[1] for p in pts) / len(pts), 2)
    centroids_iter2[zone] = (new_x, new_y)

for zone, c in centroids_iter2.items():
    members = [n for n, (z, d) in assignments_iter2.items() if z == zone]
    print(f'{zone}: members={members} → new centroid = {c}')

# No hotspot changed zone in Iteration 2 — the assignments are stable.
# The centroids also came out exactly the same as after Iteration 1,
# which tells me the algorithm has already converged.


# Task 5
# I compare the centroids from both iterations to check for convergence.
# If they are identical, the algorithm has found its final grouping
# and there is no point running more iterations.
# I also print the final zone assignments clearly so the city council
# can see which hotspots belong to each maintenance team.

print('--- Convergence Check ---')
print('Centroids after Iteration 1:', centroids_iter1)
print('Centroids after Iteration 2:', centroids_iter2)

if centroids_iter1 == centroids_iter2:
    print('Converged after 2 iterations ✅')
else:
    print('Not converged — running Iteration 3...')
    zones_iter3 = {}
    for name, (zone, d) in assignments_iter2.items():
        zones_iter3.setdefault(zone, []).append(points[name])
    centroids_iter3 = {}
    for zone, pts in zones_iter3.items():
        new_x = round(sum(p[0] for p in pts) / len(pts), 2)
        new_y = round(sum(p[1] for p in pts) / len(pts), 2)
        centroids_iter3[zone] = (new_x, new_y)
    print('Centroids after Iteration 3:', centroids_iter3)

print()
print('--- Final Zone Assignments ---')
final_zones = {}
for name, (zone, d) in assignments_iter2.items():
    final_zones.setdefault(zone, []).append(name)

for zone in sorted(final_zones.keys()):
    members = final_zones[zone]
    centroid = centroids_iter2[zone]
    print(f'{zone}: hotspots={members}, centroid={centroid}')

# Final result makes geographic sense:
# Zone1 covers the left cluster (P1, P2, P3) — all close together near coordinates (1-2, 1-3)
# Zone2 covers the bottom-right hotspots (P4, P5)
# Zone3 covers just P6 in the top-right corner
# Each maintenance team has a clearly defined area of the city to manage.


# Task 6
# P3 is at position (2,2) which sits between the left cluster and the right side.
# I wanted to check whether this boundary position caused it to switch zones
# at any point during the iterations.

p3_zone_iter1 = assignments_iter1['P3'][0]
p3_zone_iter2 = assignments_iter2['P3'][0]

print(f'P3 zone in Iteration 1: {p3_zone_iter1}')
print(f'P3 zone in Iteration 2: {p3_zone_iter2}')

if p3_zone_iter1 == p3_zone_iter2:
    print('P3 stayed in the same zone throughout.')
else:
    print('P3 switched zones between iterations!')

# P3 stayed in Zone1 throughout both iterations in my setup.
# This is because my starting centroid for Zone1 was P1(1,1), which is close to P3(2,2).
# However, P3 is a boundary point — it is closer to the left cluster but not by a huge margin.
# If I had chosen different starting centroids, Zone2 or Zone3 could have started
# closer to P3 and pulled it into a different zone.
# Points near the boundary between clusters are always at risk of switching zones
# when centroids move during iterations, which is what makes them unstable.
# The final zone of a boundary point depends heavily on where the algorithm starts.


# Task 7
# I tested what happens when all 3 centroids start in the same left area of the map.
# The idea is to show why starting centroids need to be spread out.
# With bad centroids, the algorithm has no way to explore the full map from the start.

print('--- Challenge: Bad Starting Centroids ---')

bad_centroids = {
    'Zone1': (1, 1),
    'Zone2': (1, 3),
    'Zone3': (2, 2),
}
print('Bad starting centroids:', bad_centroids)

# Iteration 1 with bad centroids
bad_assign1 = {}
for name, pt in points.items():
    best_zone = None
    best_dist = float('inf')
    for zone, c in bad_centroids.items():
        d = distance(pt, c)
        if d < best_dist:
            best_dist = d
            best_zone = zone
    bad_assign1[name] = (best_zone, round(best_dist, 2))

print('\nIteration 1 assignments (bad centroids):')
for name, (zone, d) in bad_assign1.items():
    print(f'  {name} → {zone} (distance: {d})')

bad_zones1 = {}
for name, (zone, d) in bad_assign1.items():
    bad_zones1.setdefault(zone, []).append(points[name])

bad_c2 = {}
for zone, pts in bad_zones1.items():
    bad_c2[zone] = (
        round(sum(p[0] for p in pts) / len(pts), 2),
        round(sum(p[1] for p in pts) / len(pts), 2)
    )
print('New centroids:', bad_c2)

# Iteration 2 with bad centroids
bad_assign2 = {}
for name, pt in points.items():
    best_zone = None
    best_dist = float('inf')
    for zone, c in bad_c2.items():
        d = distance(pt, c)
        if d < best_dist:
            best_dist = d
            best_zone = zone
    bad_assign2[name] = (best_zone, round(best_dist, 2))

print('\nIteration 2 assignments (bad centroids):')
for name, (zone, d) in bad_assign2.items():
    print(f'  {name} → {zone} (distance: {d})')

bad_zones2 = {}
for name, (zone, d) in bad_assign2.items():
    bad_zones2.setdefault(zone, []).append(name)

print('\nFinal zones with bad centroids:')
for zone, members in sorted(bad_zones2.items()):
    print(f'  {zone}: {members}')

# The result is clearly wrong — Zone2 ends up with only P2 as a single hotspot,
# while Zone3 gets all four right-side points (P3, P4, P5, P6) lumped together.
# This is not a useful grouping for the city council because
# one maintenance team would have almost no work and another would be overwhelmed.
# The problem is that all three starting centroids were competing for the left cluster,
# leaving the right side of the map completely uncovered at the start.
# Once the centroids finally spread out, the damage is done —
# the algorithm converges to a poor local solution instead of the true natural grouping.
# This experiment clearly shows that K-Means is only as good as its starting centroids.
```
